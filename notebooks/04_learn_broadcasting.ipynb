```python
a = torch.randn(5, 4)
b = torch.randn(3, 4)

a.shape, b.shape
```

```python
inter_x1 = torch.max(a[:, None, 0], b[None, :, 0])
inter_y1 = torch.max(a[:, None, 1], b[None, :, 1])
inter_x2 = torch.min(a[:, None, 2], b[None, :, 2])
inter_y2 = torch.min(a[:, None, 3], b[None, :, 3])

inter_x1.shape, inter_y1.shape, inter_x2.shape, inter_y2.shape
```

```python
inter_x1[0], inter_y1[0], inter_x2[0], inter_y2[0]
```

```python
intersection_coords = torch.cat((inter_x1[:, :, None], inter_y1[:, :, None], inter_x2[:, :, None], inter_y2[:, :, None]), dim=-1)
```

```python
intersection_coords.shape, intersection_coords[0]
```

```python
inter_w = (intersection_coords[..., 2] - intersection_coords[..., 0]).clamp(min=0)
inter_h = (intersection_coords[..., 3] - intersection_coords[..., 1]).clamp(min=0)

inter_w.shape, inter_w[0], inter_h.shape, inter_h[0]
```

```python
inter_w * inter_h
```
